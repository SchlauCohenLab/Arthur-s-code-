# Calculate the deepth of focus for a given lens and beam diatemer

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider
from math import pi
def DOF(wv,F,D,n):
    return (8*wv*F**2)/(pi*n*D**2)

```

```python
def updatePlot(F,D):  #F focal distance in mm, #D Beam Diameter in mm
    wv = np.linspace(400,600)*1e-6 #wavelength in mm 
    n = 1.4 #refraction index
    length = DOF(wv,F,D,n)

    plt.figure(figsize=(8, 4))
    plt.plot(wv*1e6, length, label=f'Focal = {F:.0f}, Diameter Beam = {D:.0f}')
    plt.xlabel('wavelength in nm')
    plt.ylabel('Depth of Focus in mm')
    plt.grid('On')
    plt.legend()
    plt.show()


interact(updatePlot, F=FloatSlider(min=50, max=500, step=5, value=200.0, description='Focal length (mm)'),D=FloatSlider(min=1, max=10, step=0.5, value=5, description='Beam Diameter (mm)') )
```
