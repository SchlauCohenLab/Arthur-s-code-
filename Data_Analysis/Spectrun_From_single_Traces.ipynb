This code will, from single kinetic traces, create a 3D csv file that can then be used with the main spectrum analysis code 

```python
import os
import re
import pandas as pd
from scipy.io import loadmat, savemat

# Define the folder containing the data files
data_folder = "/Users/vard/Documents/DT project/Data/250213_NaDT_Me2Styrene_air_spectrum"

# Initialize an empty dictionary to store the kinetics data
kinetics_map = {}

# Iterate over all files in the data folder
for file in os.listdir(data_folder):
    # Check if the file is a .mat file
    if file.endswith(".mat"):
        # Construct the full file path
        file_path = os.path.join(data_folder, file)

        # Extract the wavelength from the file name using a regular expression
        match = re.search(r'(\d+)', file)
        if match:
            wavelength = int(match.group(1))
        else:
            continue

        # Load the .mat file data
        mat_data = loadmat(file_path)

        # Check if the 'data' key exists in the loaded data
        if 'data' in mat_data:
            data = mat_data['data']

            # Create a DataFrame from the data
            df = pd.DataFrame(data, columns=['Time', 'TA'])

            # Insert the wavelength as a new column in the DataFrame
            df.insert(2, 'Wavelength', wavelength)

            # Store the DataFrame in the kinetics_map dictionary with the wavelength as the key
            kinetics_map[wavelength] = df

# Concatenate all DataFrames in the kinetics_map dictionary into a single DataFrame
final_df = pd.concat(kinetics_map.values(), ignore_index=True)

# Pivot the final DataFrame to have Time as the index, Wavelength as the columns, and TA as the values
final_df = final_df.pivot(index='Time', columns='Wavelength', values='TA')

# Sort the columns by wavelength
final_df = final_df.sort_index(axis=1)

# Print the first few rows of the final DataFrame
print(final_df.head())


```
